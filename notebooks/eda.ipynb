### Observe that ~73% are non-churners and ~27% are churners. Indicating class imbalance.

```python
# TotalCharges is being read as a string. Convert it to numeric

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].isnull().sum()
```

```python
# Univariate Analysis for numeric columns
df[['tenure', 'MonthlyCharges', 'TotalCharges']].describe()
```

```python
# Univariate Analysis for categorical columns
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(df[col].value_counts(), "\n")
```

## Churn Rate by Categorical Variables

```python
cat_features = ['gender', 'SeniorCitizen', 'Partner', 'Dependents',
                'PhoneService', 'MultipleLines', 'InternetService',
                'OnlineSecurity', 'Contract', 'PaymentMethod', 'PaperlessBilling']

for col in cat_features:
    churn_rate = df.groupby(col)['Churn'].apply(lambda x: (x == 'Yes').mean())
    print(churn_rate.sort_values(ascending=False), "\n")
```

```python
# Encode churn and check correlations
df['Churn_binary'] = (df['Churn'] == 'Yes').astype(int)
df[['tenure', 'MonthlyCharges', 'TotalCharges', 'Churn_binary']].corr()
```

## Churn Rate by Numerical variables

```python
df['Churn_binary'] = (df['Churn'] == 'Yes').astype(int)

df.groupby('Churn')[['tenure', 'MonthlyCharges', 'TotalCharges']].mean()
```

## Churning based on tenure segment

```python
# Creating tenure groups
df['tenure_group'] = pd.cut(df['tenure'], bins=[0, 12, 24, 48, 72],
                             labels=['0-1yr', '1-2yr', '2-4yr', '4-6yr'])
```

```python
df.groupby('tenure_group')['Churn_binary'].mean()
```

## Revenue at risk

```python
churned = df[df['Churn'] == 'Yes']
print(f"Monthly revenue at risk: ${churned['MonthlyCharges'].sum():,.0f}")
print(f"Avg monthly charge (churned): ${churned['MonthlyCharges'].mean():.2f}")
print(f"Avg monthly charge (retained): ${df[df['Churn']=='No']['MonthlyCharges'].mean():.2f}")
```

## Key Insights

- There are ~73% non-churners and ~27% churners, indicating some class imbalance.
- Month-to-month contracts churn at ~40% vs ~3% for two-year contracts
- Fiber optic internet users churn more than DSL users
- Customers without OnlineSecurity churn far more
- Churned customers have much lower tenure (avg ~18 months vs ~38 for retained) (Also indicated by negative correlation between Tenure and Churn feature)
- Churned customers pay higher monthly charges on average
- TotalCharges is lower for churned (short tenure × high monthly)
